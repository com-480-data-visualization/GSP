```python
import pandas as pd

medals = pd.read_csv('data/olympic_medals.csv')
print(medals.columns.tolist())
print(medals.dtypes)
medals.head()
```

```python
gdp = pd.read_csv('data/API_NY.GDP.MKTP.CD_DS2_en_csv_v2_94769.csv', skiprows=4)  # 4 first line are not data 

# transform column format to row format 
gdp_long = gdp.melt(
    id_vars=['Country Name', 'Country Code'],
    value_vars=[str(y) for y in range(1960, 2025)],
    var_name='Year',
    value_name='GDP'
)

gdp_long = gdp_long.dropna(subset=['GDP'])

gdp_long['Year'] = gdp_long['Year'].astype(int)
gdp_long = gdp_long[(gdp_long["Year"] <= 2016)]

gdp_long.sort_values(by= "Year") # to see when it starts (1960)
print(gdp_long.dtypes)
gdp_long.head()
```
