```python
import pandas as pd

df = pd.read_csv('../data/medals_merged_gdp_capita_log_max.csv')
df["inv_gdp_usd"] = 1 / df["gdp_usd"]
df.head()
```

```python
inv_gdp_by_country = df.groupby('country')['inv_gdp_usd'].agg(inv_gdp_usd_mean='mean', medals='count')
inv_gdp_by_country
```

```python
inv_gdp_by_country.sort_values("inv_gdp_usd_mean", ascending=False).head(10)
```

```python
# save inv_gdp_by_country to ../website/public/efficiency_by_country.csv
inv_gdp_by_country.to_csv('../website/public/efficiency_by_country.csv')
```
